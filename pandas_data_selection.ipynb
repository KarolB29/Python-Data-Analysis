```python
import pandas as pd
import os
import re
```

```python
df_raw = pd.read_csv(
    os.path.join("..", "..", "01_Dane", "ceny_produktow_renamed.csv"),
    sep=';',
    encoding='UTF-8',
    decimal='.'
)
```

```python
df_raw.head(10)
```

```python
# to localize searched product names and provinces
index = 0
for x in df_raw['product_types'].unique():
    print(index, x)
    index += 1
print('======================================')

index = 0
for y in df_raw['province'].unique():
    print(index, y)
    index += 1
```

```python
m2015 = [f"2015-{m}" for m in range(1, 13)] # creates a set of months that can be used as a filter condition
cond1 = df_raw['product_types'].unique()[4] # creates a searched name as a second filter condition
cond1
```

```python
df_carrots = df_raw.loc[
                (df_raw['product_types'].isin([cond1])) &  # filters by name
                (df_raw['date'].isin(m2015)) &  # filters only year 2015 
                (df_raw['value'] > 0)] # value of 0 is considered as error, so it will be filtered out
df_carrots.info() # check if any null left
```

```python
df_carrots['value'].describe() 
# mean price is shown below, along with other statistics values to check if they are logical
```
